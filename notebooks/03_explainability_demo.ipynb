# 03 — Explainability demo (SHAP + LIME)

Loads a demo attack flow and prints SHAP/LIME top features for the report (RQ3).

```python
import sys
from pathlib import Path

import pandas as pd

ROOT = Path("..").resolve()
sys.path.insert(0, str(ROOT))

from backend.services.pipeline import load_demo_flow, run_explain, run_prediction

demo = load_demo_flow("DDoS")
print("Ground truth:", demo["label"])
pred = run_prediction(demo["features"], db=None, persist=False)
print("Prediction:", pred["attack_type"], "risk", pred["risk_score"], pred["severity"])

shap = run_explain(demo["features"], top_k=8, method="shap")
lime = run_explain(demo["features"], top_k=8, method="lime")

print("\nSHAP:", shap["explanation"])
display(pd.DataFrame(shap["top_features"]))
print("\nLIME:", lime["explanation"])
display(pd.DataFrame(lime["top_features"]))
```
