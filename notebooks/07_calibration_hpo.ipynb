# 07 — Calibration & HPO analysis

Reads sample-based research outputs from `scripts/11_calibration_hpo_experiment.py`.

**Run first:**
```bash
python scripts/11_calibration_hpo_experiment.py
```


````python
import json
from pathlib import Path

import pandas as pd
from IPython.display import Image, Markdown, display

ROOT = Path("..").resolve()
path = ROOT / "models/trained_models/calibration_hpo_experiment.json"
fig = ROOT / "models/trained_models/figures/calibration_curve.png"

if not path.exists():
    display(Markdown("Experiment JSON missing — run `scripts/11_calibration_hpo_experiment.py`."))
else:
    report = json.loads(path.read_text(encoding="utf-8"))
    rows = []
    for name in ("baseline", "isotonic_calibrated", "randomized_search"):
        block = report[name]
        m = block.get("metrics", {})
        c = block.get("calibration", {})
        rows.append({
            "model": name,
            "f1": m.get("f1"),
            "pr_auc": m.get("pr_auc"),
            "brier": c.get("brier_score"),
            "ece": c.get("ece"),
        })
    display(Markdown("## Validation sample comparison"))
    display(pd.DataFrame(rows))
    display(Markdown(report.get("note", "")))
    if fig.exists():
        display(Image(filename=str(fig)))
````
